```python
import pandas as pd
```

```python
df = pd.read_csv("docs/trial_detail.csv")
```

```python
df.columns.values
```

```python
len(df)
```

```python
df.groupby("candidate_id").filter(lambda x: (x.trial_phase == "Phase 1").any())
```

```python
phase_col = "trial_phase" if "trial_phase" in df.columns else "phase"

by_candidate = df.groupby("candidate_id")[phase_col].agg(lambda s: {str(v).strip() for v in s.dropna()})

has_phase1 = by_candidate.apply(lambda phases: "Phase 1" in phases)
has_other_phase = by_candidate.apply(lambda phases: any(p != "Phase 1" for p in phases))

matched = has_phase1 & has_other_phase

numerator = int(matched.sum())
denominator = int(len(by_candidate))
pct = (numerator / denominator * 100) if denominator else 0.0

print(f"{numerator}/{denominator} candidate_id groups = {pct:.2f}%")

```
